```python
# Create a mapping dictionary for sorting
title_order = {title: idx for idx, title in enumerate(titles)}

# Add a temporary column for sorting (unmatched titles get a high number)
df['sort_order'] = df['Equivalent US Core Profile'].map(title_order).fillna(len(titles))

# Sort by the temporary column, then drop it
df_sorted = df.sort_values('sort_order').drop('sort_order', axis=1)


# Save to CSV
df_sorted.to_csv('sorted_output.csv', index=False)
```
